```python
import pandas as pd
import matplotlib.pyplot as plt
```

### Encoder

```python
df_encoder_training = pd.read_csv("/data/DeepMicroClass/training_checkpoints/encoder_with_projection_head_training/encoder_training_log_epoch.csv")
xticks = [f'{i}' for i in range(1, len(df_encoder_training)+1)]
df_encoder_training = df_encoder_training.set_index(pd.Index(xticks))

fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(6, 3))

ax.plot(df_encoder_training.index, df_encoder_training["loss"], label="train")
ax.plot(df_encoder_training.index, df_encoder_training["val_loss"], label="validate")
ax.set_ylabel('Supervised Contrastive Loss', fontdict={'fontsize': 10})
# ax.set_ylim(0, 15)
# ax.set_xlim(0, 10)
ax.tick_params(axis='x', labelrotation=0, labelsize=7)
ax.set_axisbelow(True)
ax.legend(prop={'size': 7}, loc='upper right')
ax.grid(True)

fig.suptitle("Loss during train and validate", fontdict={'fontsize': 10})
fig.tight_layout()
fig.savefig('loss_curve_encoder.pdf', bbox_inches='tight')
```

### Classifier

```python
df_classifier_training = pd.read_csv("/data/DeepMicroClass/training_checkpoints/classifier/classifier_training_log_epoch.csv")
xticks = [i for i in range(1, len(df_classifier_training)+1)]
df_classifier_training = df_classifier_training.set_index(pd.Index(xticks))

fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(12, 3))

ax[0].plot(df_classifier_training.index, df_classifier_training["loss"], label="train")
ax[0].plot(df_classifier_training.index, df_classifier_training["val_loss"], label="validate")
ax[0].set_ylabel('Sparse Categorical Focal Loss', fontdict={'fontsize': 10})
ax[0].tick_params(axis='x', labelrotation=0, labelsize=7)
ax[0].set_axisbelow(True)
ax[0].legend(prop={'size': 7}, loc='upper right')
ax[0].grid(True)

ax[1].plot(df_classifier_training.index, df_classifier_training["sparse_categorical_accuracy"], label="train")
ax[1].plot(df_classifier_training.index, df_classifier_training["val_sparse_categorical_accuracy"], label="validate")
ax[1].set_ylabel('Sparse Categorical Accuracy', fontdict={'fontsize': 10})
ax[1].set_ylim(0, 1)
ax[1].tick_params(axis='x', labelrotation=0, labelsize=7)
ax[1].set_axisbelow(True)
ax[1].legend(prop={'size': 7}, loc='lower right')
ax[1].grid(True)

fig.suptitle("Loss and Accuracy during train and validate", fontdict={'fontsize': 10})
fig.tight_layout()
fig.savefig('loss_accuracy_curve_classifier.pdf', bbox_inches='tight')
```
